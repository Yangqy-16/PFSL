**Calculate Ranking Metrics**

```python
import pandas as pd
from sklearn.metrics import precision_score, recall_score
from sklearn.metrics import ndcg_score
import numpy as np

# 假设df是你的DataFrame，包含四列：0(gene1), 1(gene2), 2, prob

def calculate_metrics(df: pd.DataFrame, k_values: list[int] = [10, 20]) -> dict[str, float]:
    # 按prob降序排序
    df_sorted = df.sort_values(by='prob', ascending=False)

    # 计算precision@k, recall@k, ndcg@k
    metrics = {}
    for k in k_values:
        top_k = df_sorted.head(k)

        # Precision@k
        precision = precision_score(top_k['2'], (top_k['prob'] >= 0.5).astype(int))

        # Recall@k
        recall = recall_score(top_k['2'], (top_k['prob'] >= 0.5).astype(int))

        # NDCG@k
        relevance = top_k['2'].values  # 这里假设label为1为正样本，0为负样本
        ndcg = ndcg_score([relevance], [top_k['prob'].values])

        metrics[f'Precision@{k}'] = precision
        metrics[f'Recall@{k}'] = recall
        metrics[f'NDCG@{k}'] = ndcg

    return metrics
```

```python
slbench_logit_path = '/data/qingyuyang/ESM4SL/data/SLKB/slbench/case' #specific
slgnnct_logit_path = '/home/qingyuyang/SLGNNCT/results'
slgnnct_input_path = '/data/qingyuyang/ESM4SL/data/SLKB/slgnnct/case' #specific
mvgcn_logit_path = '/data/qingyuyang/MVGCNiSL/results'
esm4sl_logit_path = '/data/qingyuyang/ESM4SL/output/pfnew' #'/data/qingyuyang/ESM4SL/data/attn_results'
esm4sl_input_path = '/data/qingyuyang/ESM4SL/data/SLKB/case' #specific

# rank_save_path = '/data/qingyuyang/ESM4SL/all_methods_rank_count'
# os.makedirs(rank_save_path, exist_ok=True)

def df_cal_mean_std(df: pd.DataFrame) -> list[str]:
    mean_values = df.iloc[0:5].mean()
    std_values = df.iloc[0:5].std()
    return [f"{mean:.4f} ({std:.4f})" for mean, std in zip(mean_values, std_values)]

def cal_all_methods_rank(cl: str, scene: str) -> pd.DataFrame:
    column_names = ['Precision@10', 'Precision@20', 'Recall@10', 'Recall@20', 'NDCG@10', 'NDCG@20']
    slmgae_df = pd.DataFrame(0, index=range(6), columns=column_names)
    kg4sl_df = pd.DataFrame(0, index=range(6), columns=column_names)
    ptgnn_df = pd.DataFrame(0, index=range(6), columns=column_names)
    nsf4sl_df = pd.DataFrame(0, index=range(6), columns=column_names)
    slgnnct_df = pd.DataFrame(0, index=range(6), columns=column_names)
    if os.path.exists(f'{mvgcn_logit_path}/{cl}/{scene}'):
        mvgcn_df = pd.DataFrame(0, index=range(6), columns=column_names)
    esm4sl_df = pd.DataFrame(0, index=range(6), columns=column_names)

    for fold in range(5):
        slmgae_logit = pd.read_csv(f'{slbench_logit_path}/{cl}/{scene}/SLMGAE/sl_test_{fold}.csv')
        slmgae_metrics = calculate_metrics(slmgae_logit)
        slmgae_df.iloc[fold] = [slmgae_metrics['Precision@10'], slmgae_metrics['Precision@20'], slmgae_metrics['Recall@10'],
                                slmgae_metrics['Recall@20'], slmgae_metrics['NDCG@10'], slmgae_metrics['NDCG@20']]

        kg4sl_logit = pd.read_csv(f'{slbench_logit_path}/{cl}/{scene}/KG4SL/sl_test_{fold}.csv')
        kg4sl_metrics = calculate_metrics(kg4sl_logit)
        kg4sl_df.iloc[fold] = [kg4sl_metrics['Precision@10'], kg4sl_metrics['Precision@20'], kg4sl_metrics['Recall@10'],
                               kg4sl_metrics['Recall@20'], kg4sl_metrics['NDCG@10'], kg4sl_metrics['NDCG@20']]
        
        ptgnn_logit = pd.read_csv(f'{slbench_logit_path}/{cl}/{scene}/PTGNN/sl_test_{fold}.csv')
        ptgnn_metrics = calculate_metrics(ptgnn_logit)
        ptgnn_df.iloc[fold] = [ptgnn_metrics['Precision@10'], ptgnn_metrics['Precision@20'], ptgnn_metrics['Recall@10'],
                               ptgnn_metrics['Recall@20'], ptgnn_metrics['NDCG@10'], ptgnn_metrics['NDCG@20']]
        
        nsf4sl_logit = pd.read_csv(f'{slbench_logit_path}/{cl}/{scene}/NSF4SL/sl_test_{fold}.csv')
        nsf4sl_metrics = calculate_metrics(nsf4sl_logit)
        nsf4sl_df.iloc[fold] = [nsf4sl_metrics['Precision@10'], nsf4sl_metrics['Precision@20'], nsf4sl_metrics['Recall@10'],
                                nsf4sl_metrics['Recall@20'], nsf4sl_metrics['NDCG@10'], nsf4sl_metrics['NDCG@20']]

        slgnnct_logit_raw = pd.read_csv(f'{slgnnct_logit_path}/{cl}_{scene}/fold_{fold+1}/test.csv').rename({'2': 'prob'}, axis=1)
        # slgnnct_gt = pd.read_csv(f'{slgnnct_input_path}/{cl}/{scene}/sl_test_{fold}.csv')
        slgnnct_logit = slgnnct_logit_raw.rename({'label': '2', 'logit': 'prob'}, axis=1) #pd.merge(slgnnct_gt, slgnnct_logit_raw, on=['0', '1'])[['0', '1', '2', 'prob']]
        slgnnct_metrics = calculate_metrics(slgnnct_logit)
        slgnnct_df.iloc[fold] = [slgnnct_metrics['Precision@10'], slgnnct_metrics['Precision@20'], slgnnct_metrics['Recall@10'],
                                 slgnnct_metrics['Recall@20'], slgnnct_metrics['NDCG@10'], slgnnct_metrics['NDCG@20']]

        if os.path.exists(f'{mvgcn_logit_path}/{cl}/{scene}'):
            mvgcn_logit = pd.read_csv(f'{mvgcn_logit_path}/{cl}/{scene}/{fold}.csv').rename({'label': '2'}, axis=1)
            mvgcn_metrics = calculate_metrics(mvgcn_logit)
            mvgcn_df.iloc[fold] = [mvgcn_metrics['Precision@10'], mvgcn_metrics['Precision@20'], mvgcn_metrics['Recall@10'],
                                   mvgcn_metrics['Recall@20'], mvgcn_metrics['NDCG@10'], mvgcn_metrics['NDCG@20']]

        version_num = len(os.listdir(f'{esm4sl_logit_path}/{cl}/{scene}/fold_{fold}/csv_log')) - 1 #_wrs
        esm4sl_logit_raw = pd.read_csv(f'{esm4sl_logit_path}/{cl}/{scene}/fold_{fold}/csv_log/version_{version_num}/test_logits.csv') #_wrs
        esm4sl_gt = pd.read_csv(f'{esm4sl_input_path}/{cl}/{scene}/sl_test_{fold}.csv')
        esm4sl_logit = pd.merge(esm4sl_gt, esm4sl_logit_raw, left_on=['0', '1'], right_on=['gene1', 'gene2'])[['0', '1', '2', 'probs']]
        esm4sl_logit.rename({'probs': 'prob'}, axis=1, inplace=True)
        esm4sl_metrics = calculate_metrics(esm4sl_logit)
        esm4sl_df.iloc[fold] = [esm4sl_metrics['Precision@10'], esm4sl_metrics['Precision@20'], esm4sl_metrics['Recall@10'],
                                esm4sl_metrics['Recall@20'], esm4sl_metrics['NDCG@10'], esm4sl_metrics['NDCG@20']]
    
    combined_df = pd.DataFrame('', index=range(7), columns=column_names)
    combined_df.index = ['SLMGAE', 'KG4SL', 'PTGNN', 'NSF4SL', 'SLGNNCT', 'MVGCN-iSL', 'ESM4SL']
    combined_df.iloc[0] = df_cal_mean_std(slmgae_df)
    combined_df.iloc[1] = df_cal_mean_std(kg4sl_df)
    combined_df.iloc[2] = df_cal_mean_std(ptgnn_df)
    combined_df.iloc[3] = df_cal_mean_std(nsf4sl_df)
    combined_df.iloc[4] = df_cal_mean_std(slgnnct_df)
    if os.path.exists(f'{mvgcn_logit_path}/{cl}/{scene}'):
        combined_df.iloc[5] = df_cal_mean_std(mvgcn_df)
    combined_df.iloc[6] = df_cal_mean_std(esm4sl_df)
    # combined_df.to_csv(f'{rank_save_path}/{cl}_{scene}.csv')
    return combined_df
```

```python
def cal_all_methods_all_metrics(cl: str, scene: str) -> pd.DataFrame:
    cls_metrics = count_all_methods_specific(cl, scene)
    rank_metrics = cal_all_methods_rank(cl, scene)
    all_metrics = pd.merge(cls_metrics, rank_metrics, left_index=True, right_index=True)
    
    auc_mean = [float(i.split()[0]) if i != '' else '' for i in list(all_metrics['AUC'])]
    aupr_mean = [float(i.split()[0]) if i != '' else '' for i in list(all_metrics['AUPR'])]
    f1_mean = [float(i.split()[0]) if i != '' else '' for i in list(all_metrics['F1'])]
    bacc_mean = [float(i.split()[0]) if i != '' else '' for i in list(all_metrics['BACC'])]
    sum4 = [a1 + a2 + a3 + a4 for a1, a2, a3, a4 in zip(auc_mean, aupr_mean, f1_mean, bacc_mean)]
    # all_metrics['Sum4'] = sum4

    pre10_mean = [float(i.split()[0]) if i != '' else '' for i in list(all_metrics['Precision@10'])]
    pre20_mean = [float(i.split()[0]) if i != '' else '' for i in list(all_metrics['Precision@20'])]
    rec10_mean = [float(i.split()[0]) if i != '' else '' for i in list(all_metrics['Recall@10'])]
    rec20_mean = [float(i.split()[0]) if i != '' else '' for i in list(all_metrics['Recall@20'])]
    sum6 = [a1 + a2 + a3 for a1, a2, a3 in zip(sum4, pre20_mean, rec20_mean)]
    all_metrics['Sum6'] = sum6

    all_metrics.to_csv(f'{save_path}/{cl}_{scene}.csv')
    return all_metrics
```
