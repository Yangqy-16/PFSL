**SLGNNCT: change all genes back to name, to match its KG database**

```python
from utils import *

root_path = f'{root}/SLKB'
```

```python
slgnnct_specific_save_path = f'{root_path}/slgnnct/specific_1:1'
os.makedirs(slgnnct_specific_save_path, exist_ok=True)

input_path = '/data/qingyuyang/SL/PFSL/data/SLKB/specific_1:1'

for cl in os.listdir(input_path):  # cell_lines ['JURKAT']
    for scene in ['C1', 'C2', 'C3', 'Tail']: #'AKT1'
        if not os.path.exists(f'{input_path}/{cl}/{scene}'): #specific_save_path
            continue
        save_path = f'{slgnnct_specific_save_path}/{cl}/{scene}'
        os.makedirs(save_path, exist_ok=True)

        raw_path = f'{input_path}/{cl}/{scene}'  #table_save_path
        for fn in tqdm(os.listdir(raw_path)):
            sl = pd.read_csv(f'{raw_path}/{fn}')
            old_len = len(sl)
            sl[GENE1_COLNAME] = sl[GENE1_COLNAME].map(gid2gn)
            sl[GENE2_COLNAME] = sl[GENE2_COLNAME].map(gid2gn)
            sl.dropna(inplace=True)
            sl.rename({GENE1_COLNAME: 'gene1', GENE2_COLNAME: 'gene2', CLASSLABEL_COLNAME: 'label'}, axis=1, inplace=True)
            new_len = len(sl)
            assert old_len == new_len
            sl.to_csv(f'{save_path}/{fn}', index=False)
```
