```python
import pandas as pd
import os
import pickle

input_dir = '/data/siyutao/SL_recom_BCB/SL_data'
output_dir = '/data/qingyuyang/ESM4SL/data/ryf'
os.makedirs(output_dir, exist_ok=True)

with open('/data/qingyuyang/ESM4SL/data/mapping/name2id.pkl', 'rb') as f:
    gn2gid = pickle.load(f)
gn2gid['CCDC84'] = '338657'
```

```python
def transform_df(df: pd.DataFrame) -> pd.DataFrame:
    ori_len = len(df)
    df.rename({'gene_1': '0', 'gene_2': '1', 'label': '2'}, axis=1, inplace=True)
    df.replace({"2": {'Not SL': 0, 'SL': 1}}, inplace=True)
    df['0'] = df['0'].map(gn2gid)
    df['1'] = df['1'].map(gn2gid)
    df.dropna(inplace=True, ignore_index=True)
    assert len(df) == ori_len
    return df
```

```python
for folder in ['30033366']: #os.listdir(input_dir)
    for scene in os.listdir(f'{input_dir}/{folder}'):
        sce_in_dir = f'{input_dir}/{folder}/{scene}'
        sce_out_dir = f'{output_dir}/{folder}/{scene}'
        os.makedirs(sce_out_dir, exist_ok=True)
        
        for i in range(5):
            train = pd.read_csv(f'{sce_in_dir}/sl_tr_{i}.csv')
            train = transform_df(train)
            train.to_csv(f'{sce_out_dir}/sl_train_{i}.csv', index=False)

            val = pd.read_csv(f'{sce_in_dir}/sl_val_{i}.csv')
            val = transform_df(val)
            val.to_csv(f'{sce_out_dir}/sl_val_{i}.csv', index=False)

            test = pd.read_csv(f'{sce_in_dir}/sl_te_{i}.csv')
            test = transform_df(test)
            test.to_csv(f'{sce_out_dir}/sl_test_{i}.csv', index=False)
```
