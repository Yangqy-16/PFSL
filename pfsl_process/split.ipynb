```python
import os
import pandas as pd
import pickle

# with open('/data/qingyuyang/SL/PFSL/data/mapping/name2id.pkl', 'rb') as f:
#     gn2gid = pickle.load(f)

slkb = pd.read_csv('/data/qingyuyang/SL/PFSL/data/SLKB/SLKB_processed.csv')
# slkb['0'] = slkb['gene_1'].map(gn2gid).astype(int)
# slkb['1'] = slkb['gene_2'].map(gn2gid).astype(int)

root = '/data/qingyuyang/SL/PFSL/data/SLKB/specific'
out_root = '/data/qingyuyang/SL/PFSL/data/SLKB/specific_1:1'

for cl in os.listdir(root):
    if cl.endswith('.csv') or cl == 'RPE1':
        continue
    print(cl)

    slkb_cl = slkb[slkb['context'] == cl]

    for scene in os.listdir(f'{root}/{cl}'):
        if '_' in scene:
            continue
        print(scene)

        out_dir = f'{out_root}/{cl}/{scene}'
        os.makedirs(out_dir, exist_ok=True)

        for fn in os.listdir(f'{root}/{cl}/{scene}'):
            print(fn)

            original = pd.read_csv(f'{root}/{cl}/{scene}/{fn}')
            merged = pd.merge(original, slkb_cl, left_on=['0', '1'], right_on=['gene1', 'gene2'])

            true_samples = merged[merged['2'] == 1]
            false_samples = merged[merged['2'] == 0]
            false_keep = false_samples.sort_values(by=['score'], ascending=False)[:len(true_samples)]

            final = pd.concat([true_samples, false_keep], axis=0, ignore_index=True)[['0', '1', '2', 'score']]
            final.to_csv(f'{out_dir}/{fn}', index=False)
            
    #         break
    #     break
    # break
```

```python
import pandas as pd
import os
import pickle

with open('/data/qingyuyang/SL/PFSL/data/mapping/gid2uid.pkl', 'rb') as f:
    gid2uid = pickle.load(f)

original_root = '/data/qingyuyang/SL/PFSL/data/SLKB/specific_1:1'
new_root = '/data/qingyuyang/SL/PFSL/data/slbenchmark/specific_1:1'
os.makedirs(new_root, exist_ok=True)

for cl in os.listdir(original_root):
    for scene in os.listdir(f'{original_root}/{cl}'):
        for fn in os.listdir(f'{original_root}/{cl}/{scene}'):
            data = pd.read_csv(f'{original_root}/{cl}/{scene}/{fn}')

            original_len = len(data)
            data['0'] = data['0'].map(gid2uid)
            data['1'] = data['1'].map(gid2uid)
            data = data.dropna()
            new_len = len(data)
            assert original_len == new_len

            os.makedirs(f'{new_root}/{cl}/{scene}', exist_ok=True)
            data.to_csv(f'{new_root}/{cl}/{scene}/{fn}', index=False)
```
