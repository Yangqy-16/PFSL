```python
import pandas as pd
import os

root = '/data/qingyuyang/SL/PFSL/data/SLKB/specific/JURKAT'

new_slkb = pd.read_csv('/data/qingyuyang/SL/PFSL/data/SLKB/SLKB_processed.csv')
new_slkb = new_slkb[new_slkb['context'] == 'JURKAT']

output_root = '/data/qingyuyang/SL/PFSL/data/SLKB/regression'

# for cl in os.listdir():
#     if cl.endswith('.csv'):
#         continue

for scene in ['C3']: #os.listdir(root) f'{root}/{cl}'
    for dataset in os.listdir(f'{root}/{scene}'):
        origin_processed = pd.read_csv(f'{root}/{scene}/{dataset}')

        new_processed = pd.merge(origin_processed, new_slkb, left_on=['0', '1'], right_on=['gene1', 'gene2'])
        new_processed = new_processed[['0', '1', 'score']].rename({'score': '2'}, axis=1)

        os.makedirs(f'{output_root}/{scene}', exist_ok=True)
        new_processed.to_csv(f'{output_root}/{scene}/{dataset}', index=False)
```
